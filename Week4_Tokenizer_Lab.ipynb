```python
from transformers import AutoTokenizer

print("Tokenizer library imported successfully!")
```

```python
TOKENIZERS = {
    "BPE - GPT-2": "gpt2",
    "WordPiece - Multilingual BERT": "bert-base-multilingual-cased",
    "SentencePiece - XLM-R": "xlm-roberta-base"
}

print("Three tokenizers defined successfully.")

for name, model in TOKENIZERS.items():
    print(name, ":", model)
```

```python
TEXTS = {
    "English": "Artificial intelligence is changing the world.",
    "Telugu": "కృత్రిమ మేధస్సు ప్రపంచాన్ని మారుస్తోంది.",
    "Hindi": "कृत्रिम बुद्धिमत्ता दुनिया को बदल रही है।"
}

for language, text in TEXTS.items():
    print(language, ":", text)
```

```python
loaded_tokenizers = {}

for display_name, model_name in TOKENIZERS.items():
    print(f"Loading {display_name}...")

    loaded_tokenizers[display_name] = (
        AutoTokenizer.from_pretrained(model_name)
    )

print("\nAll three tokenizers loaded successfully!")
```

```python
def compare_tokenizers(loaded_tokenizers):

    for language, text in TEXTS.items():

        print("\n" + "=" * 75)
        print(f"LANGUAGE: {language}")
        print(f"INPUT TEXT: {text}")
        print("=" * 75)

        for tokenizer_name, tokenizer in loaded_tokenizers.items():

            encoded = tokenizer(
                text,
                add_special_tokens=True,
                truncation=False
            )

            token_ids = encoded["input_ids"]

            tokens = tokenizer.convert_ids_to_tokens(token_ids)

            print(f"\n{tokenizer_name}")
            print("Tokens      :", tokens)
            print("Token IDs   :", token_ids)
            print("Token count :", len(token_ids))


compare_tokenizers(loaded_tokenizers)
```

```python
import pandas as pd

results = []

for language, text in TEXTS.items():

    for tokenizer_name, tokenizer in loaded_tokenizers.items():

        encoded = tokenizer(
            text,
            add_special_tokens=True,
            truncation=False
        )

        tokens = tokenizer.convert_ids_to_tokens(
            encoded["input_ids"]
        )

        results.append({
            "Language": language,
            "Tokenizer": tokenizer_name,
            "Token Count": len(encoded["input_ids"]),
            "Tokens": " | ".join(tokens)
        })

df = pd.DataFrame(results)

display(df)
```

```python
df.to_csv("tokenizer_comparison_results.csv", index=False)

from google.colab import files
files.download("tokenizer_comparison_results.csv")
```
